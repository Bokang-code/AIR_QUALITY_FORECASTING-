# 02 - Exploratory Data Analysis (EDA)
This notebook performs basic EDA on the processed air quality dataset and saves figures to `outputs/figures/`.

```python
import pandas as pd
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

sns.set(style='whitegrid')
```

```python
project_root = Path.cwd().resolve()
for candidate in [project_root, *project_root.parents]:
    if (candidate / 'src').exists() or (candidate / 'requirements.txt').exists():
        project_root = candidate
        break

processed_path = project_root / 'data' / 'processed' / 'processed_air_quality_data.csv'
print('Looking for processed file at', processed_path)

if not processed_path.exists():
    print('Processed file not found at', processed_path)
else:
    df = pd.read_csv(processed_path, parse_dates=['date_utc'])
    display(df.head())
```

```python
# Basic stats
print(df.info())
print(df.describe())
print('\nMissing values:')
print(df.isna().sum())
```

```python
# Time-series example for PM2.5 (parameter name may vary)
param = 'pm25'
if 'parameter' in df.columns and param in df['parameter'].unique():
    df_param = df[df['parameter'] == param].set_index('date_utc').resample('D')['value'].mean()
    fig = plt.figure(figsize=(10,4))
    df_param.plot(title=f'Daily mean {param}')
    Path('outputs/figures').mkdir(parents=True, exist_ok=True)
    fig.savefig('outputs/figures/ts_pm25.png')
    print('Saved outputs/figures/ts_pm25.png')
else:
    print(f'No measurements for {param}')
```

```python
# Histogram of values
fig = plt.figure(figsize=(6,4))
sns.histplot(df['value'].dropna(), bins=50)
fig.savefig('outputs/figures/value_hist.png')
print('Saved outputs/figures/value_hist.png')
```
